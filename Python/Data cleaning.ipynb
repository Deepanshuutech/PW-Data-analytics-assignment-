```python
import pandas as pd
import numpy as np 

df = pd.read_csv('healthcare_data_cleaning_dataset.csv')
df['Admission_Date'] = pd.to_datetime(df['Admission_Date'])
```

```python
#Q1

df.isnull().sum()
(df.isnull().sum() / len(df)) * 100
```

```python
#Q2

df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].isnull().sum()
```

```python
#Q3
df['Treatment_Cost'].skew() 
df['Treatment_Cost'] = df['Treatment_Cost'].fillna(df['Treatment_Cost'].median())
```

```python
#Q4 
df.duplicated().sum() 
df.duplicated(subset=['Patient_ID']).sum() 
df = df.drop_duplicates() 
df = df.drop_duplicates(subset=['Patient_ID'], keep='first')
```

```python
len(df)
```

```python
#Q5

df[(df['Age'] > 100) | (df['Age'] < 0)]
```

```python
df.loc[(df['Age'] > 100) | (df['Age'] < 0), 'Age'] = np.nan
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
#Q6

Q1 = df['Treatment_Cost'].quantile(0.25)
Q3 = df['Treatment_Cost'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
outliers = df[(df['Treatment_Cost'] < lower) | (df['Treatment_Cost'] > upper)]
print(len(outliers))
```

```python
#Q7

p5 = df['Treatment_Cost'].quantile(0.05)
p95 = df['Treatment_Cost'].quantile(0.95)

print("5th percentile:", p5)
print("95th percentile:", p95)

df['Treatment_Cost_Capped'] = df['Treatment_Cost'].clip(lower=p5, upper=p95)

print("Before capping -> Min:", df['Treatment_Cost'].min(), " Max:", df['Treatment_Cost'].max())
print("After capping  -> Min:", df['Treatment_Cost_Capped'].min(), " Max:", df['Treatment_Cost_Capped'].max())

print("Total rows:", len(df))
```

```python
#Q8

df['Treatment_Cost_Log'] = np.log1p(df['Treatment_Cost'])

df['Treatment_Cost'].skew()
df['Treatment_Cost_Log'].skew()
```

```python
#Q9

df = df.sort_values('Admission_Date')

df['Admission_Date'] = df['Admission_Date'].ffill()
df['Admission_Date'] = df['Admission_Date'].bfill()

print("Missing dates after:", df['Admission_Date'].isnull().sum())
```
