```python
import pandas as pd

df = pd.read_csv("Advanced_pandas_sales_dataset.csv")
df.head()
```

```python
#Q1

df.head(10)
df.info()
print("Shape (rows, columns):", df.shape)
print(df.dtypes)
```

```python
#Q2

# 1. Sales > 3500
high_sales = df[df["Sales"] > 3500]
high_sales

# 2. East region + Sales > 2500
east_high = df[(df["Region"] == "East") & (df["Sales"] > 2500)]
east_high
```

```python
#Q3

top_customers = df.groupby("Customer")["Sales"].sum().sort_values(ascending=False).head(5)
top_customers
```

```python
#Q4

region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(region_sales)

print("Highest revenue region:", region_sales.idxmax())
```

```python
#Q5

product_qty = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)
print(product_qty)

print("Most popular product:", product_qty.idxmax())
```

```python
#Q6

region_category_sales = df.groupby(["Region", "Category"])["Sales"].sum()
region_category_sales
```

```python
#Q7

pivot_table = pd.pivot_table(df, index="Region", columns="Product", values="Sales", aggfunc="sum")
pivot_table
```

```python
print("Region with most laptop sales:", pivot_table["Laptop"].idxmax())
```

```python
#Q8

category_profit = df.groupby("Category")["Profit"].mean().sort_values(ascending=False)
print(category_profit)

print("Most profitable category:", category_profit.idxmax())
```

```python
#Q9

discount_profit = df.groupby("Discount")["Profit"].mean().sort_values()
discount_profit
```

```python
print("Correlation between Discount and Profit:", df["Discount"].corr(df["Profit"]))
```

```python
#Q10

# Step 1: Filter
filtered = df[df["Sales"] > 3000]

# Step 2 & 3: Group by Region + Product, total sales
grouped = filtered.groupby(["Region", "Product"])["Sales"].sum().reset_index()

# Step 4: Reshape into pivot table
final_pivot = grouped.pivot(index="Region", columns="Product", values="Sales")
final_pivot
```
