```python
import pandas as pd 
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

df = pd.read_csv('employee_productivity_dataset.csv')
df.head()
```

```python
#Q1

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Hours_Worked_Per_Week'] = df['Hours_Worked_Per_Week'].fillna(df['Hours_Worked_Per_Week'].median())
df['Performance_Score'] = df['Performance_Score'].fillna(df['Performance_Score'].mean())

df.isnull().sum()
df.head()
```

```python
#Q2

le = LabelEncoder()

df['Gender_Encoded'] = le.fit_transform(df['Gender'])
df['Department_Encoded'] = le.fit_transform(df['Department'])

df[['Gender', 'Gender_Encoded', 'Department', 'Department_Encoded']].head()
```

```python
#Q3

df_before_cols = df.shape[1]

df = pd.get_dummies(df, columns=['Work_Mode', 'Location'])

df_after_cols = df.shape[1]

print("Columns before:", df_before_cols)
print("Columns after:", df_after_cols)
print("New columns created:", df_after_cols - df_before_cols)
df.head()
```

```python
#Q4

minmax = MinMaxScaler()

df[['Salary_MinMax', 'Hours_MinMax']] = minmax.fit_transform(df[['Salary', 'Hours_Worked_Per_Week']])

df[['Salary', 'Salary_MinMax', 'Hours_Worked_Per_Week', 'Hours_MinMax']].head()
```

```python
#Q5

standard = StandardScaler()

df[['Age_Standard', 'Projects_Standard']] = standard.fit_transform(df[['Age', 'Projects_Completed']])

df[['Age', 'Age_Standard', 'Projects_Completed', 'Projects_Standard']].head()
```

```python
#Q6

df['Salary_MinMax_Compare'] = MinMaxScaler().fit_transform(df[['Salary']])
df['Salary_Standard_Compare'] = StandardScaler().fit_transform(df[['Salary']])

df[['Salary', 'Salary_MinMax_Compare', 'Salary_Standard_Compare']].head()
```

```python
#Q7

from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler, OneHotEncoder

categorical_cols = ['Gender', 'Department']  # ya jo bhi raw categorical columns bache hain
numerical_cols = ['Age', 'Salary', 'Hours_Worked_Per_Week', 'Projects_Completed']

num_pipeline = Pipeline(steps=[
    ('scaler', StandardScaler())
])

preprocessor = ColumnTransformer(transformers=[
    ('num', num_pipeline, numerical_cols),
    ('cat', OneHotEncoder(), categorical_cols)
])
```

```python
#Q8

transformed_data = preprocessor.fit_transform(df)

print("Shape of final dataset:", transformed_data.shape)
print(transformed_data[:5])   # pehle 5 rows dekho
```

```python
#Q9 Why is scaling important?

"""
Different columns in a dataset have very different ranges — for example, Salary might range from 40,000 to 100,000,
while Age only ranges from 20 to 60. If we don't scale the data,
models like KNN, SVM, or Linear Regression end up treating the column with bigger numbers as more important,
even though it might not actually be. Scaling brings all columns to a similar range, so the model trains fairly and gives better,
more accurate results
"""
```

```python
#Q10 Why do we convert categorical data into numerical form?

"""
Machine learning models can only understand numbers — they can't directly work with text values like "Male", "Female", "IT", or "HR".
So we use Label Encoding or One-Hot Encoding to turn these text categories into numbers,
which allows the model to actually process and learn from that column.
If we skip this step, the model will either throw an error or won't be able to use that column at all.
"""
```
